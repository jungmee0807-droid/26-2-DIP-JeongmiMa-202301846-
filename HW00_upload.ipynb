- Image file의 metadata를 확인하기

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image, ImageOps
from PIL.ExifTags import TAGS
```

```python
raw = Image.open('20260711_191127.jpg')          # 원본 (메타데이터 포함)
I = np.array(ImageOps.exif_transpose(raw))       # Orientation 정보 반영해서 바로 세우기

print('데이터타입 :', I.dtype)
print('배열 shape :', I.shape)
print('공간해상도 :', f'{I.shape[1]} x {I.shape[0]} pixels (가로 x 세로), 채널 {I.shape[2]}개')
print('최소값     :', I.min())
print('최대값     :', I.max())
```

```python
I = Image.open('20260711_191127.jpg')
exifdata = I.getexif()
for tag_id in exifdata:
    tag = TAGS.get(tag_id, tag_id)
    data = exifdata.get(tag_id)
    print(f"{tag:25}: {data}")
```

```python
plt.figure(figsize=(6, 8))
plt.imshow(I)
plt.title('A place I’d love to go back to', fontsize=15)
plt.axis('off')
plt.show()
```

```python
import pydicom
```

```python
dcm = pydicom.dcmread('IM-0001-0001.dcm')
```

```python
print("modality:",dcm.Modality)
print("hospital:",dcm.InstitutionName)
print("voxelsize:",dcm.PixelSpacing)
print("slice thickness:",dcm.SliceThickness)
```

```python
f = dcm.pixel_array
print(f.shape)
print("Lmin : ",np.min(f))
print("Lmax : ",np.max(f))
```

```python
plt.figure(dpi=128)
plt.imshow(f,vmin=np.min(f),vmax=np.max(f),cmap='gray')
plt.colorbar()
plt.axis('off')
plt.title('Time of Flight MR angiography')
plt.show()
```

```python
f_inv = np.max(f)-f
plt.figure(dpi=128)
plt.imshow(f_inv,vmin=np.min(f_inv),vmax=np.max(f_inv),cmap='gray')
plt.colorbar()
plt.axis('off')
plt.title('Time of Flight MR angiography (inverted)')
plt.show()
```
